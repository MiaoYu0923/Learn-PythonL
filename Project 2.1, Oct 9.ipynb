```python
print("Letter Grade Converter")
print()
#Need to judge whether you need to do it again or not, use c to mark this
c = "y"
while c == "y":
    #Ask the user to give the grade
    grade = int(input("Enter numerical grade: "))
    #To set the grade range
    if 88 <= grade and grade <= 100:
        print("Entered numerical grade: " + str(grade))
        print("Letter grade: A")
    elif grade >= 80:
        print("Entered numerical grade: " + str(grade))
        print("Letter grade: B")
    elif grade >= 67:
        print("Entered numerical grade: " + str(grade))
        print("Letter grade: C")
    elif grade >= 60:
        print("Entered numerical grade: " + str(grade))
        print("Letter grade: D")
    elif grade >= 0:
        print("Entered numerical grade: " + str(grade))
        print("Letter grade: F")
    else:
        print("Please enter a valid grade.")

    c = input("Continue? (y/n): ")
    print()
    print("Continue? (y/n): " + str(c))
    print()
    if c == "n":
        print("Bye!")
        break





```
