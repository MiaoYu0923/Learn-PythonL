```python
print("Table of Powers")
print()
#enter start number and stop number
start_num = int(input("Enter start number: "))
stop_num = int(input("Enter end number: "))
print()
print("Number" + "  " + "Squared" + "  " + "Cubed")
print("=====" + "  " + "=====" + "  " + "======")
for n in range(start_num, stop_num+1):
    print(str(n) + "     " + str(n*n) + "    " + str(n*n*n) )
```
