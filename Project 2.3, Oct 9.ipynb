```python
print("Change Calculator")
print()
#need to set up a loop
c = "y"
while c == "y":
    #type the number
    num = int(input("Enter a number: "))
    print("Enter number of cents (0-99): " + str(num))
    print()
    #calculate quarters, dimes, nickles and pennies
    quarters = num // 25
    num -= quarters * 25
    dimes = num // 10
    num -= dimes * 10
    nickles = num // 5
    num -= nickles * 5
    pennies = num
    #print all these numbers
    print("Quarters: " + str(quarters))
    print("Dimes: " + str(dimes))
    print("Nickles: " + str(nickles))
    print("Pennies: " + str(pennies))
    print()
    #check if you want to quit the loop or not
    c = input("Do you want to continue? (y/n): ")
    if c == "y":
        print("Continue? (y/n): y")
        print()
    elif c == "n":
        print("Continue? (y/n): n")
        print()
        print("Bye!")
        break
    else:
        print("Please enter y or n only.")



```
