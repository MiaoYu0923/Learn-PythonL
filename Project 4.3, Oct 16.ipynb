```python
# define a function to convert feet to meters
def feet_to_meters(feet):
    meters = feet * 0.3048
    return meters
# define a function to convert meters to feet
def meters_to_feet(meters):
    feet = meters / 0.3048
    return feet
#define a main function
def main():
    print("Feet and Meters Converter")
    # need to see whether they want to continue to calculate it again, so set a loop
    to_continue = True
    while to_continue:
        print()
        print("Conversions Menu:")
        print("a. Feet to Meters")
        print("b. Meters to Feet")
        # make a choice
        choice = input("Select a or b")
        print("Select a conversion (a/b): " + choice)
        # to idenfity which function should be applied
        # if they want feet to meters
        if choice == "a":
            feet = float(input("Please input feet: "))
            meters = feet_to_meters(feet)
            print()
            print("Enter feet: " + str(feet))
            print(str(meters) + " meters")
        # if they want meters to feet
        elif choice == "b":
            meters = float(input("Please input meters: "))
            feet = round(meters_to_feet(meters),2)
            print()
            print("Enter meters: " + str(meters))
            print(str(feet) + " feet")
        print()
        # to see if they want another conversion
        decision = input("Would you like to continue? (y/n)")
        print("Would you like to perform another conversion? (y/n): " + decision)
        if decision == "y":
            to_continue = True
        elif decision == "n":
            to_continue = False
            print()
            print("Thanks, bye!")
            break
        else:
            print("Invalid input. Please try again.")

if __name__ == "__main__":
    main()


```
