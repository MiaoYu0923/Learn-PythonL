```python
# define a function that converts miles to feet
def miles_to_feet(m):
    return m * 5280
# define a main function
def main():
    print("Hike Calculator")
    print()
    #type the miles
    m = float(input("How many miles did you walk? "))
    #apply for the "miles_to_feet" function to calculate feet
    feet = miles_to_feet(m)
    print("How many miles did you walk?: " + str(m))
    print("You walked " + str(feet) + " feet.")
if __name__ == '__main__':
    main()

```
