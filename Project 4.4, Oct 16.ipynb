```python
# calculate the tax, the tax rate is 6%
def tax(total):
    return total * 0.06
# calculate the total of the items, n is for how many items there are
def total_amount(n):
    total = 0
    print("ENTER ITEMS (ENTER 0 TO END)")
    for i in range(0,n):
        # type each item cost one by one
        item = float(input("Enter the item cost: "))
        print("Cost of item: " + str(item))
        total += item
    return total


# define display
def display():
    print("Sales Tax Calculator")
    print()

# define the main function
def main():
    display()
    # to set a loop to see whether it needs to continue
    to_continue = True
    while to_continue:
        n = int(input("Enter the number of items: "))
        total =total_amount(n)
        print("Total: \t\t\t" + str(total))
        sales_amount = round(tax(total),2)
        print("Sales tax: \t\t" + str(sales_amount))
        final_amount = round((total + sales_amount), 2)
        print("Total after tax: " + str(final_amount))
        print()
        # to receive the signal whether needs to start a new round
        decision = input("Would you like to continue? (Y/N): ")
        print("Again? (y/n): " + str(decision))
        if decision == "y":
            to_continue = True
        elif decision == "n":
            to_continue = False
            print()
            print("Thanks, bye!")
            break
        else:
            print("Please enter y or n only.")
if __name__ == "__main__":
    main()




```
