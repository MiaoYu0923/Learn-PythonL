```python
#import random
import random
#define display
def display():
    print("Dice Roller")
    print()

#define roll function so it will show number from 1 to 6
def roll():
    value = random.randint(1,6)
    return value
#define a role dice function
def roll_dice():
    die1 = roll()
    die2 = roll()
    total = die1 + die2
    print("Die 1: ", die1)
    print("Die 2: ", die2)
    print("Total: ", total)
    if total == 2:
        print("Snake eyes!")
    elif total == 12:
        print("Boxcars!")
    print()

def main():
    display()
    # to decide if you want to roll the dice
    decision = input("Would you like to roll? (y/n): ")
    print("Roll the dice? (y/n): " + decision)
    print()
    while decision == "y":
        roll_dice()
        decision = input("Roll again? (y/n): ")
        print("Roll again? (y/n): " + decision)
        print()


#if started as the main module, call the main function
if __name__ == "__main__":
    main()
```
