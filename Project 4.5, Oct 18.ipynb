```python
import random
def roll():
    value = random.randint(1,6)
    return value
def display():
    print("Dice Roller")
    print()
def dice_roll():
    dice1 = roll()
    dice2 = roll()
    total = dice1 + dice2
    print("Dice 1:", dice1)
    print("Dice 2:", dice2)
    print("Total:", total)

    if total == 6:
        print("Boxcars!")
    elif total == 2:
        print("Snake eyes!")
    print()
def main():
    display()
    # to see if a roll dice is needed
    decision = input("Roll the dice? (y/n): ")
    print("Roll the dice? (y/n): " + decision)
    while decision == "y":
        dice_roll()
        decision = input("Roll the dice? (y/n): ")
        print("Roll again? (y/n): " + decision)
#if started with as the main module, call the main function
if __name__ == "__main__":
    main()


```
