```python
print("Tip Calculator")
print()
#input your cost of meal
meal_cost = float(input("Enter meal cost: "))
print("Cost of meal: " + str(meal_cost))
print()
tip_percentage = 0
# need these three tip percentages
for t in range(15,30,5):
    tip_percentage = round(t / 100, 2)
    #show the tip percentage
    print(str(int(tip_percentage*100)) + "%")
    #calculate the tip amount
    tip_amount = round(meal_cost * tip_percentage, 2)
    #show the tip amount
    print("Tip amount: " + str(tip_amount))
    #calculate the total amount
    total_amount = round(meal_cost + tip_amount, 2)
    #show the total amount
    print("Total amount: " + str(total_amount))
    print()













```
