```python
print("================================")
print("Shipping Calculator")
print("================================")
#set a loop
continue_sign = "y"
while continue_sign == "y":
    #offer the ordered cost
    ordered_cost = float(input("Enter the cost of your order: "))
    #to decide the shipping cost based on ordered cost
    if ordered_cost <= 0:
        print("Cost of items ordered: " + str(ordered_cost))
        print("You must enter a positive number. Please try again.")
        continue
    elif ordered_cost < 30:
        shipping_cost = 5.95
    elif ordered_cost <= 49.99:
        shipping_cost = 7.95
    elif ordered_cost <= 74.99:
        shipping_cost = 9.95
    elif ordered_cost > 75:
        shipping_cost = 0

    #name the total cost
    total_cost = round(ordered_cost + shipping_cost,2)
    print("Cost of items ordered: " + str(ordered_cost))
    print("Shipping cost: " + str(shipping_cost))
    print("Total cost: " + str(total_cost))
    print()
    continue_sign = input("Continue? (y/n) ")
    if continue_sign == "n":
        print("Continue? (y/n): n")
        print("================================")
        print("Bye!")
        break
    elif continue_sign == "y":
        print("Continue? (y/n): y")
        print("================================")
    else:
        print("Please enter y or n only.")
```
